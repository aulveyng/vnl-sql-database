```python
# ============================================
# Volleyball Nations League Analytics Platform
# Model Comparison
# ============================================

# =====================================================
# Import Libraries
# =====================================================

import pandas as pd

# =====================================================
# Model Performance Summary
# =====================================================

results = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Random Forest",
        "Gradient Boosting"
    ],

    "Accuracy": [
        0.556,
        0.611,
        0.527
    ],

    "Precision": [
        0.50,
        0.56,
        0.48
    ],

    "Recall": [
        0.69,
        0.63,
        0.75
    ],

    "F1 Score": [
        0.58,
        0.59,
        0.59
    ],

    "ROC-AUC": [
        0.61,
        0.69,
        0.59
    ]
})

# =====================================================
# Display Results
# =====================================================

print("=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

display(results)

# =====================================================
# Rank Models
# =====================================================

ranking = results.sort_values(
    by=["ROC-AUC", "Accuracy", "F1 Score"],
    ascending=False
).reset_index(drop=True)

print("\nOverall Ranking")
display(ranking)

# =====================================================
# Select Best Model
# =====================================================

best_model = ranking.iloc[0]

print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(f"Model      : {best_model['Model']}")
print(f"Accuracy   : {best_model['Accuracy']:.3f}")
print(f"Precision  : {best_model['Precision']:.3f}")
print(f"Recall     : {best_model['Recall']:.3f}")
print(f"F1 Score   : {best_model['F1 Score']:.3f}")
print(f"ROC-AUC    : {best_model['ROC-AUC']:.3f}")

# =====================================================
# Save Results
# =====================================================

results.to_csv(
    "../outputs/model_comparison.csv",
    index=False
)

print("\nModel comparison saved to outputs/model_comparison.csv")
```
